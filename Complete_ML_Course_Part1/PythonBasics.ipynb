## 🖨️ Print Function
- `print()` displays output to the screen.  
- Strings must be enclosed in matching quotes (single or double).  

```python
print("Machine Learning")                 # String literal
print('Projects')                         # String literal (single quotes)
print("Machine" + " " + "Learning")       # Concatenation
print(8)                                  # Number
print(8 + 3)                              # Arithmetic
```

## 🔤 Basic Data Types
- `int`: whole numbers (e.g., `10`, `-7`)  
- `float`: decimal numbers (e.g., `3.14`)  
- `str`: text (e.g., `"hello"`)  
- `bool`: logical values (`True`, `False`)  
- `complex`: real + imaginary part (e.g., `1 + 3j`)  

```python
print(type(8))                                    # int
print(type(5.3))                                  # float
print(type("Hello"))                              # str
print(type(True))                                 # bool
print(type(1 + 3j))                               # complex
```

## 📦 Variables and Constants
- A variable stores a value and can be reassigned.  
- A constant is named in ALL_CAPS by convention; reassignment is still possible but discouraged.  

```python
superhero = "Iron Man"                      # Variable assignment
print(superhero)                            # Iron Man

superhero = "Thor"                          # Reassign variable
print(superhero)                            # Thor

PI = 3.14159                                # Constant by convention (ALL_CAPS)
print("Pi =", PI)                           # Pi = 3.14159

PI = 10                                     # Reassignment allowed (not recommended)
print("New Pi =", PI)                       # New Pi = 10
```

## 📝 Multiple Assignment
- Assign multiple values in one line.  
- Assign the same value to several variables.  

```python
hero1, hero2, hero3 = "Iron Man", "Captain America", "Batman"  # Tuple unpacking
print(hero1, hero2, hero3)                                     # Iron Man Captain America Batman

x = y = z = 23                                                  # Chain assignment
print(x, y, z)                                                  # 23 23 23
```

## 🔄 Type Conversion
- Convert between types using `int()`, `float()`, and `str()`.  
- Useful when numeric data is stored as strings.  

```python
num = 5                                      # int
print(num, type(num))                        # 5 <class 'int'>

num = float(num)                             # int → float
print(num, type(num))                        # 5.0 <class 'float'>

value = int(7.9)                             # float → int (truncates)
print(value, type(value))                    # 7 <class 'int'>

total = int("10") + int("20")                # strings → ints
print(total)                                 # 30
```

## 🧵 Strings
- Quotes: single or double (must match).  
- Concatenation: `+`  
- Repetition: `*`  
- Slicing: `s[start:stop:step]` (start at 0, stop excluded).  

```python
print("Machine Learning")                       # Double quotes
print('Machine Learning')                       # Single quotes

print("Hello" * 3)                              # Repetition
print("Machine" + " " + "Learning")             # Concatenation

s = "programming"                               # Indexes 0..10
print(s[1:5])                                   # rogr  (1..4)
print(s[0:10:2])                                # pormrga (every 2nd char)
```

## 🧰 Lists (mutable, ordered, allow duplicates)
- Enclose elements in `[ ]`.  
- Can mix types.  
- Support indexing and many methods.  

```python
my_list = [2, 3, 1.8, "english", True]        # Mixed types
print(my_list, type(my_list))                 # [...] <class 'list'>

my_list.append(6)                             # Append element
print(my_list)                                # Element added at end

print(my_list[0])                             # Indexing → 2
print(my_list[2])                             # Indexing → 1.8

lst1 = [1,2,3,4,5,1,2,3]                      # Duplicates allowed
print(lst1)                                   # [1, 2, 3, 4, 5, 1, 2, 3]
print(len(lst1))                              # Length → 8

lst2 = []                                     # Empty list + append
lst2.append(5)
print(lst2)                                   # [5]

lst2 = [2,3,1.8,"english",True,6]
del lst2[2]                                   # Delete by index
print(lst2)                                   # 1.8 removed

l3 = [1,2,3,4,5]                              # Concatenate lists
l4 = [6,7,8,9,10]
l5 = l3 + l4
print(l5)                                     # [1..10]
```

## 🔒 Tuples (immutable, ordered)
- Enclose elements in `( )`.  
- Support indexing; cannot be modified.  

```python
t1 = (2,3,4,5)
print(t1, type(t1))                           # (2, 3, 4, 5) <class 'tuple'>

t2 = (1, 3.5, "machine learning", False)      # Mixed types
print(t2)

lst = [3,4,5,6]
t = tuple(lst)                                # List → Tuple
print(lst, t)

print(t[0], t[1])                             # Indexing → 3 4
# t.append(6)                                 # AttributeError (immutable)
print(len(t))                                 # 4
```

## 🧮 Sets (mutable, unordered, unique elements)
- Enclose elements in `{ }`.  
- No indexing; duplicates removed automatically.  

```python
s = {1,2,3,4,5}
print(s, type(s))                             # {1, 2, 3, 4, 5} <class 'set'>

# s[0]                                        # TypeError (no indexing)

lst5 = [4,5,6,7,7,6]
x = set(lst5)                                 # List → Set (deduplicate)
print(x)                                      # {4, 5, 6, 7}

s3 = {1,2,3,4,5,1,2,3}
print(s3)                                     # {1, 2, 3, 4, 5}
```

## 📚 Dictionaries (mutable, key:value pairs)
- Enclose in `{ }` with `key: value`.  
- Access values by key.  

```python
person = {"name": "David", "age": 30, "country": "India"}
print(person, type(person))                         # {...} <class 'dict'>

print(person["name"])                               # David
print(person["age"])                                # 30
print(person["country"])                            # India

d2 = {"name": "David", "age": 30, "name": "Alice"}  # Duplicate key
print(d2)                                           # {'name': 'Alice', 'age': 30}
```

## ➗ Operators
Python provides several types of operators. These allow performing arithmetic, comparisons, logical checks, and more.

### Arithmetic Operators  
`+  -  *  /  **  %` perform basic math operations.

```python
num1, num2 = 20, 10

sum_val  = num1 + num2    # 20 + 10 → 30
diff_val = num1 - num2    # 20 - 10 → 10
prod_val = num1 * num2    # 20 * 10 → 200
quot_val = num1 / num2    # 20 / 10 → 2.0 (float division)
exp_val  = num1 ** 2      # 20 ** 2 → 400 (power)
mod_val  = num1 % 3       # 20 % 3 → 2 (remainder)

print(sum_val, diff_val, prod_val, quot_val, exp_val, mod_val)
```

### Assignment Operators  
`=  +=  -=  *=  /=  **=  %=` are used to assign and update values.

```python
a = 5        # assign → 5
print(a)

a += 5       # a = a + 5 → 10
print(a)

a -= 2       # a = a - 2 → 8
print(a)

a *= 3       # a = a * 3 → 24
print(a)

a /= 4       # a = a / 4 → 6.0
print(a)

a **= 2      # a = a ** 2 → 36.0
print(a)

a %= 7       # a = a % 7 → 1.0
print(a)
```

### Comparison Operators  
`==  !=  >  <  >=  <=` compare values and return `True` or `False`.

```python
a, b = 5, 10
print(a == b)   # equal? → False
print(a != b)   # not equal? → True
print(a >  b)   # greater than? → False
print(a <  b)   # less than? → True
print(a >= b)   # greater or equal? → False
print(a <= b)   # less or equal? → True
```

### Logical Operators  
`and  or  not` combine conditions.

```python
x = 10
print((x > 5) and (x < 20))   # True and True → True
print((x > 20) or  (x > 5))   # False or True → True
print(not (x > 5))            # not True → False
```

### Identity Operators  
`is`, `is not` check whether two variables refer to the *same object* in memory (not just equal values).

```python
x, y = 5, 5
print(x is y)       # may be True (small ints are cached)
print(x is not y)   # opposite of 'is'
print(x == y)       # value equality → True
```

### Membership Operators  
`in`, `not in` check if a value exists inside a collection (list, set, string, etc.).

```python
a, b = 5, 10
c = [1, 2, 3, 4, 5]

print(a in c)        # 5 in list? → True
print(b in c)        # 10 in list? → False
print(a not in c)    # 5 not in list? → False
print(b not in c)    # 10 not in list? → True
```
