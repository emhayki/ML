## 🔀 If, Elif, and Else
Conditional statements let Python make decisions.

- `if`: runs only if the condition is **True**
- `elif`: extra condition(s) checked if previous ones are False
- `else`: runs if none of the above conditions are True

**Indentation rule:** Each block under `if`, `elif`, or `else` must be indented by 4 spaces.

### ✅ Two-way choice (if--else)

```python
a, b = 15, 25   # two numbers to compare

if a > b:
    print("a is greater")
else:
    print("b is greater")   # runs because 25 > 15
```

### 🔄 Multi-way choice (if--elif--else)

```python
a, b, c = 15, 25, 30   # three numbers

if a > b and a > c:
    print("a is greatest")
elif b > c:
    print("b is greatest")
else:
    print("c is greatest")   # runs because 30 is largest
```

### 🪜 Nested if (less common)
Nested `if` means putting one `if` inside another. This works but can get hard to read.

```python
a, b, c = 20, 40, 60   # three numbers

if a > b:
    if a > c:
        print("a is greatest")
    else:
        print("c is greatest")
else:
    if b > c:
        print("b is greatest")
    else:
        print("c is greatest")   # runs because 60 is largest
```

## 🔁 Loops
Python provides two main ways to repeat code: **for loops** and **while loops**.

- `for`: use when the number of iterations is known or when looping over a sequence.
- `while`: use when repetition depends on a condition that changes during execution.

### 🔢 For Loops
Repeat a block of code a fixed number of times or directly iterate through a sequence.

#### Fixed number of repetitions (using `range`)

```python
for i in range(5):   # runs from 0 → 4
    print("Iteration:", i)
```

#### Iterate over each element in a sequence

```python
numbers = [50, 100, 150, 200]
for n in numbers:
    print("Value:", n)
```

### ♾️ While Loops
Repeat as long as a condition is **True**.

#### Run until condition becomes false

```python
i = 0
while i < 5:        # loop continues while i < 5
    print(i)
    i += 1
```

#### Condition false at the start

```python
i = 5
while i < 3:        # false immediately → loop skipped
    print(i)
    i += 1
```
