# Data Text Vectorization using TF–IDF

Text data, such as messages, reviews, or articles, cannot be directly used by most machine learning models.  
- It must first be converted into numbers that represent how important each word is in the text.  
    - This process is called **TF–IDF**, which stands for **Term Frequency – Inverse Document Frequency**.

### What is TF–IDF?

TF–IDF gives each word a score that shows how useful it is for understanding a document.  
It looks at two things:

1. **Term Frequency (TF):** how often a word appears in a single document.  
2. **Inverse Document Frequency (IDF):** how rare the word is across all documents in the dataset.

Words that appear often in one document but not in others get a **high score**, while very common words such as “the”, “and”, or “is” get a **low score**.  

### 1) Load a Built-in Text Dataset

A small sample from the **20 Newsgroups dataset** (available in scikit-learn) is used for this example. The dataset contains thousands of real-world text posts from online discussion groups, each belonging to a specific topic.

To make the concept easier to understand, only two topics are selected:

- `sci.space` → posts about space, astronomy, and NASA missions  
- `rec.sport.baseball` → posts about baseball games, teams, and players  

    - Each text post represents a **document**, and together they form the **corpus**.  
    - By comparing the words used in each topic, it becomes easier to see how **TF–IDF** identifies important terms that distinguish one subject from another.

```python
# Import libraries
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Load two small text categories
data = fetch_20newsgroups(subset="train", categories=["sci.space", "rec.sport.baseball"], remove=("headers", "footers", "quotes"))

# View a few example texts and their categories
print("Sample categories:", data.target_names)

# show the first 400 characters of one document
print("\nExample text snippet:\n")
print(data.data[0][:400])
```

### 2) Convert Text into TF–IDF Features

The **TfidfVectorizer** from scikit-learn is used to turn raw text documents into a form that machine learning models can understand (numbers).  

This process creates a **TF–IDF matrix**, which represents how important each word is within a collection of documents.

- Each **row** represents one document (for example, a single article or post).  
- Each **column** represents a unique word that appears in the dataset.  
- Each **cell value** shows how important that word is in that document, considering both its frequency in the document and its rarity across all documents.  

**Main parameters:**
- `stop_words="english"` removes very common words like *the*, *is*, or *and*, which do not add meaning.  
- `lowercase=True` converts all text to lowercase, ensuring consistency (so *Space* and *space* are treated the same).  
- `ngram_range=(1, 1)` uses only single words (unigrams).  

After applying the vectorizer, the resulting matrix has dimensions showing:
- The number of documents (rows)  
- The number of unique terms (columns)  
This gives a numerical representation of text data that can be used for modeling or pattern analysis.

```python
# Initialize the TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    stop_words="english",  # remove common words like "the", "is"
    lowercase=True,        # convert everything to lowercase
    ngram_range=(1, 1)     # use only single words (unigrams)
)

# Fit on the dataset and transform the text
X_tfidf = vectorizer.fit_transform(data.data)

# Create a DataFrame for easy viewing
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=vectorizer.get_feature_names_out())

# Show the shape and a small part of the table
print("TF-IDF matrix shape:", tfidf_df.shape)
tfidf_df.iloc[:5, :10]
```

### 3) Interpret TF–IDF Values

After converting text into numerical TF–IDF features, the next step is to understand what these values represent.

1. Choose one document (for example, one article or forum post) from the dataset.  
2. Look at the **top words** with the highest TF–IDF scores in that document.  
3. These words are the most distinctive (they appear frequently in this document but are less common in others).

**Why this matters:**
- Words with high TF–IDF scores help reveal the *main idea* or *topic* of a document.  
- They capture what makes each text unique compared to the rest of the dataset.

This analysis shows how TF–IDF automatically highlights the vocabulary that best describes the topic of each document.

```python
# Get feature names (words)
words = vectorizer.get_feature_names_out()

# Look at the first document's top words
doc_index = 0
scores = X_tfidf[doc_index].toarray().ravel()
top_indices = scores.argsort()[::-1][:10]

print("Top 10 important words in document 0:\n")
for i in top_indices:
    print(f"{words[i]}: {scores[i]:.4f}")
```
