# 🌊 Seaborn

Seaborn is a high-level **data visualization library** built on top of Matplotlib.  
It provides simple tools for creating **beautiful and informative statistical plots** with minimal code.  

Seaborn works seamlessly with **Pandas DataFrames**, making it ideal for analyzing datasets used in data science and machine learning.

**This notebook covers:**
1. Setup and imports  
2. Built-in datasets  
3. Relational plots  
4. Scatter plots  
5. Count and bar plots  
6. Distribution plots and correlation heatmaps  

## 1) Setup and Imports

```python
import seaborn as sns                 # main visualization library
import matplotlib.pyplot as plt       # supports advanced plot control
import numpy as np                    # for numerical data
import pandas as pd                   # for data manipulation
```

## 2) Built-in Datasets
Seaborn includes several example datasets for practice.  
Each dataset loads directly as a Pandas DataFrame, making exploration simple.

```python
# Load a few built-in datasets
tips = sns.load_dataset('tips')        # restaurant bills and tips
iris = sns.load_dataset('iris')        # flower measurements
titanic = sns.load_dataset('titanic')  # Titanic passenger data

# Display a few rows of each
print("=== Tips Dataset (restaurant bills and tips) ===")
display(tips.head())

print("\n=== Iris Dataset (flower measurements) ===")
display(iris.head())

print("\n=== Titanic Dataset (passenger survival data) ===")
display(titanic.head())
```

## 3) Relational Plot (Faceting and Encodings)
A relational plot shows how two numerical variables relate while encoding other features using **color**, **style**, or **size**.

```python
sns.relplot(
    data=tips,                   # Dataset to use
    kind='scatter',              # Type of relational plot ('scatter' or 'line')
    x='total_bill',              # X-axis: total bill amount
    y='tip',                     # Y-axis: tip amount
    col='time',                  # Create separate plots (facets) for Lunch and Dinner
    hue='smoker',                # Color points by smoker status
    style='smoker',              # Change marker style for smoker/non-smoker
    size='size',                 # Vary point size by table size
    sizes=(20, 200),             # Define min and max marker sizes
    height=4,                    # Height (in inches) of each subplot
    aspect=1.1                   # Aspect ratio (width = 1.1 × height)
)

plt.show()
```

## 4) Scatter Plots (Comparing Features)
Scatter plots are used to examine relationships between two continuous features.  

```python
# -------------------------------------------------------------------
# Exploring Relationships in the Iris Dataset
# The 'iris' dataset contains measurements of flower parts for 3 species: Setosa, Versicolor, and Virginica.
# -------------------------------------------------------------------

# 1) Sepal Length vs Petal Length
# Each point represents a flower, colored by species.
sns.scatterplot(
    data=iris,
    x='sepal_length',      # X-axis: sepal (outer petal) length
    y='petal_length',      # Y-axis: petal (inner petal) length
    hue='species',         # Color points by species (categorical hue)
    palette='viridis',     # Optional: color palette for better distinction
    s=60,                  # Marker size
    alpha=0.8              # Slight transparency to reduce overlap
)
plt.title('Iris Dataset (Sepal Length vs Petal Length)')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.grid(True, alpha=0.3)
plt.show()


# 2) Sepal Length vs Petal Width
# This plot shows how petal width varies with sepal length across species.
sns.scatterplot(
    data=iris,
    x='sepal_length',      # X-axis: sepal length
    y='petal_width',       # Y-axis: petal width
    hue='species',         # Color points by species
    palette='coolwarm',    # Different color palette for variety
    s=60,
    alpha=0.8
)
plt.title('Iris Dataset (Sepal Length vs Petal Width)')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.grid(True, alpha=0.3)
plt.show()
```

## 5) Count Plots (Displaying Category Frequencies)
A count plot displays how many observations fall into each category.  
It is often used for categorical or classification data such as passenger class or survival status.

```python
# -------------------------------------------------------------------
# Titanic Dataset (Exploring Passenger Class and Survival)
# The Titanic dataset includes demographic and survival data for passengers aboard the RMS Titanic.
# -------------------------------------------------------------------

# 1) Passenger Class Distribution
# Count how many passengers were in each ticket class (First, Second, Third)
sns.countplot(
    data=titanic,
    x='class',            # X-axis: passenger ticket class
    hue='class',
    palette='pastel',     # Soft color palette for clarity
    edgecolor='black'     # Outline bars for better visibility
)
plt.title('Titanic (Passenger Count by Ticket Class)')
plt.xlabel('Ticket Class')
plt.ylabel('Number of Passengers')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()


# 2) Survival Counts
# Shows how many passengers survived (1) vs did not survive (0)
sns.countplot(
    data=titanic,
    x='survived',         # 0 = did not survive, 1 = survived
    hue='survived',
    palette='coolwarm',
    edgecolor='black'
)
plt.title('Titanic (Survival Counts (0 = No, 1 = Yes))')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```

## 6) Bar Plot with Hue (Comparing Subgroups)
A bar plot shows the **average** value of a variable for each category.  
When the variable is binary (e.g., `survived`), the bar height represents the **proportion of positive outcomes**.  
Adding a `hue` allows subgroup comparisons such as gender and passenger class.

```python
# -------------------------------------------------------------------
# Titanic Dataset (Survival Rate by Sex and Class)
# This bar plot shows the average survival rate (proportion survived)
# for each sex (male/female), split by passenger class (First, Second, Third).
# -------------------------------------------------------------------

sns.barplot(
    data=titanic,
    x='sex',                   # Categorical variable on the X-axis
    y='survived',              # Numeric variable to summarize (mean survival rate)
    hue='class',               # Color bars by passenger class
    errorbar=None,             # Removes confidence interval bars (modern syntax)
    palette='Set2',            # Color palette for distinction
    edgecolor='black'          # Outlines bars for clarity
)

plt.title('Titanic (Survival Rate by Sex, Split by Class)')
plt.xlabel('Sex')
plt.ylabel('Proportion Survived')
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(title='Class')
plt.show()
```

## 7) Distribution and Correlation Heatmap

- The **California Housing dataset**, available as a built-in dataset in **Scikit-learn**, contains real-world housing data collected from California districts.  
- It shows relationships between numeric features such as **median income**, **house age**, **average rooms**, and **median house value**.  
- This section uses it to visualize **feature distributions** and a **correlation heatmap**, helping identify how variables relate to each other.

```python
from sklearn.datasets import fetch_california_housing

# Load dataset and rename target column
cal = fetch_california_housing(as_frame=True)
house = cal.frame.rename(columns={'MedHouseVal': 'Price'})
house.head()
```

### Distribution of House Prices
The `histplot` function displays how often each value appears in the dataset. Adding `kde=True` overlays a smooth density curve for a clearer shape.

```python
sns.histplot(house['Price'], bins=30, kde=True)
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```

### Correlation Matrix Heatmap
- The correlation matrix helps identify relationships between numeric variables.  
    - Values near **+1** indicate a strong positive relationship, while values near **-1** indicate a strong negative relationship.

```python
corr = house.corr(numeric_only=True)

plt.figure(figsize=(10, 10))
sns.heatmap(
    corr, annot=True, fmt='.1f', cmap='coolwarm',
    square=True, cbar=True, linewidths=.5, linecolor='white'
)
plt.title('Correlation Heatmap of California Housing Data')
plt.show()
```

**Reading the Heatmap:**
- Lighter or negative cells indicate negative correlation (for example, higher crime → lower price).  
- Darker or positive cells indicate positive correlation (for example, more rooms → higher price).  
- This view helps identify features that most strongly influence the target variable.
