## 🛠️ Functions in Python
A function is a **reusable block of code** defined with `def`.
- Can take parameters
- May return values
- Improves clarity and reusability

### ✨ Basic definition and call

```python
def greet(name):
    """Return a friendly greeting."""
    return f"Hello, {name}!"

print(greet("Ada"))   # Hello, Ada!
```

### 🔧 Parameters
- Positional and keyword
- Defaults (make args optional)
- `*args` collects extra positional
- `**kwargs` collects extra keyword

```python
def area(width, height=1):
    return width * height

def demo(*args, **kwargs):
    return args, kwargs

print(area(5))            # 5
print(area(5, 3))         # 15
print(demo(1, 2, x=10))   # ((1, 2), {'x': 10})
```

### 📤 Returning values (tuple unpacking)

```python
def stats(nums):
    total = sum(nums)
    n = len(nums)
    return total, n, total / n

s, n, mean = stats([2, 4, 9])
print(s, n, mean)    # 15 3 5.0
```

### 📖 Docstrings (in-code help)

```python
def normalize(text):
    """Return lowercase text without spaces at edges."""
    return text.strip().lower()

help(normalize)
```

⚠️ **Indentation:** After a colon `:`, indent by 4 spaces. The indented lines are the function body.

### 📍 Scope (variables inside/outside functions)

```python
counter = 0

def tick():
    global counter      # rebinds global variable
    counter += 1

def outer():
    x = 0
    def inner():
        nonlocal x      # rebinds variable in outer function
        x += 1
        return x
    return inner()
```

### 🧮 Worked Example: Factorial

```python
def factorial(n):
    """Return n! for n >= 0. Raise ValueError if negative."""
    if n < 0:
        raise ValueError("n must be >= 0")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))   # 120
```

```python
import math
print(math.factorial(5))  # 120 (built-in alternative)
```
