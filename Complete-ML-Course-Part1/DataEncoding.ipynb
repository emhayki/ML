# Data Encoding

Machine learning models can only process numerical data, but many datasets include categorical or text-based features such as gender, species, or education level.  
- These values must be converted into numeric form so that algorithms can interpret them correctly.  
This process is known as **data encoding**.

### Why Encoding Matters
Encoding ensures that categorical variables are represented in a way that models can interpret meaningfully:
- Converts **non-numeric labels** (like “Male”, “Female”, “Other”) into integers.
- Allows algorithms such as **Logistic Regression**, **SVM**, or **Neural Networks** to process input features uniformly.
- Prevents errors that occur when raw text is passed to a model.

### Common Encoding Techniques
| Method | Description | Example |
|:-------|:-------------|:--------|
| **Label Encoding** | Assigns an integer to each unique category. | `{"Male": 0, "Female": 1}` |
| **One-Hot Encoding** | Creates a new column for each category using binary indicators. | `"Color" → [Red: 1, Blue: 0, Green: 0]` |
| **Ordinal Encoding** | Applies when categories have a meaningful order. | `"Low" → 1, "Medium" → 2, "High" → 3` |

### Example 1: Penguin Dataset

The Penguin dataset is a small, clean dataset included with Seaborn.  
- It contains measurements for three penguin species **Adelie**, **Chinstrap**, and **Gentoo** along with attributes such as island, sex, and body measurements.  

- In this example, the goal is to **convert the text labels in the `species` column into numeric form** using `LabelEncoder`.  

```python
# -----------------------------------
# Import required libraries
# -----------------------------------
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

# -----------------------------------
# Load dataset
# -----------------------------------
penguins = sns.load_dataset("penguins").dropna(subset=["species"])

# -----------------------------------
# Preview dataset
# -----------------------------------
print("Dataset overview:")
print(penguins.head(), "\n")

print("Dataset shape (rows, columns):", penguins.shape, "\n")

# -----------------------------------
# Apply Label Encoding to 'species'
# -----------------------------------
encoder = LabelEncoder()
penguins["species_encoded"] = encoder.fit_transform(penguins["species"])

# -----------------------------------
# Verify encoding result
# -----------------------------------
print("Encoded dataset preview:")
print(penguins[["species", "species_encoded"]].head(), "\n")
```

### Example 2: Iris Dataset

The Iris dataset is one of the most famous datasets in machine learning.  
- It includes 150 samples of iris flowers from three species **setosa**, **versicolor**, and **virginica** along with four numeric features describing each flower (sepal and petal length and width).

- The objective here is to **encode the `species` column** into numerical values using `LabelEncoder`, allowing the dataset to be used for classification algorithms.

```python
# -----------------------------------
# Import necessary libraries
# -----------------------------------
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# -----------------------------------
# Load the Iris dataset
# -----------------------------------
iris = sns.load_dataset("iris")

# -----------------------------------
# Preview dataset
# -----------------------------------
print("Preview of the dataset:")
print(iris.head(), "\n")

print("Dataset shape (rows, columns):", iris.shape, "\n")

# -----------------------------------
# Display class distribution
# -----------------------------------
print("Count of each iris species (before encoding):")
print(iris["species"].value_counts(), "\n")

# -----------------------------------
# Apply Label Encoding to 'species'
# -----------------------------------
encoder = LabelEncoder()
iris["species_encoded"] = encoder.fit_transform(iris["species"])

# -----------------------------------
# Verify encoding results
# -----------------------------------
print("Encoded dataset preview:")
print(iris[["species", "species_encoded"]].head(), "\n")
```
